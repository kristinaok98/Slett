```python
# Importerer math for å bruke pi og kvadratrot
import math

# Lager en funksjon som beregner areal og ytre omkrets
def figur(a, b):

    # Beregner arealet av trekanten
    areal_trekant = (a * b) / 2

    # Beregner arealet av halvsirkelen
    areal_halvsirkel = (math.pi * (a / 2)**2) / 2

    # Legger sammen arealene
    areal = areal_trekant + areal_halvsirkel

    # Beregner den skrå siden i trekanten ved hjelp av Pytagoras
    skrå_side = math.sqrt(a**2 + b**2)

    # Beregner buen på halvsirkelen
    bue = math.pi * a / 2

    # Beregner ytre omkrets
    omkrets = b + skrå_side + bue

    # Returnerer areal og omkrets
    return areal, omkrets


# Ber skrive inn verdiene for a og b
a = float(input('Skriv inn verdien for a: '))
b = float(input('Skriv inn verdien for b: '))

# Kaller på funksjonen og lagrer resultatene
areal, omkrets = figur(a, b)

# Skriver ut resultatene
print('Arealet av figuren er:', areal)
print('Den ytre omkretsen av figuren er:', omkrets)
```

```python
# Importerer nødvendige biblioteker
import numpy as np
import matplotlib.pyplot as plt

# Lager 200 jevnt fordelte punkter mellom -10 og 10
x = np.linspace(-10, 10, 200)

# Beregner funksjonsverdiene
f = -x**2 - 5

# Plotter funksjonen
plt.plot(x, f)

# Legger til tittel og navn på aksene
plt.title('Grafen til f(x) = -x^2 - 5')
plt.xlabel('x')
plt.ylabel('f(x)')

# Viser koordinatnettet
plt.grid()
```
